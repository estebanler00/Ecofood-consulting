```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

"""REQUEST HTTP COMUN, SIMPLEMENTE SE PIDE UNA PAGINA"""
response =  requests.get("https://www.guiaoleo.com/buenos-aires/restaurants")
html = response.text
# print(html) muestra el texto fuente de la pagina

"""AHORA TOCA PARCEAR EL HTML PARA PODER NAVEGARLO"""
from bs4 import BeautifulSoup
soup = BeautifulSoup(html, "html.parser")
#Con BeautifulSoup podés:
# encontrar etiquetas (<h1>, <div>, etc.)
# sacar texto
# buscar por clases o IDs
# recorrer la estructura


"""ENCONTRAR QUE ETIQUETA Y QUE CLASE ENVUELVE"""
#div class="flex items-start mb-1"


"""USO ESA ETIQUETA Y CLASE PARA BUSCAR LOS DATOS"""
resto_nombre = soup.find_all("div", class_="flex items-start mb-1")
for resto in resto_nombre:
    nombre = resto.find("span").get_text(strip=True)
    # print(nombre)


"""ENCONTRAR LA FICHA DE CADA RESTAURANTE"""
resto_ficha = soup.find_all("div", class_="flex items-center flex-wrap gap-x-2 gap-y-1 text-sm text-gray-600 dark:text-gray-400")


"""OTRO FOR PARA SACAR LOS DATOS DE CADA RESTAURANTE"""
for resto in resto_ficha:
    spans = resto.find_all("span")
    if len(spans) >= 4:
        localizacion = spans[0].get_text(strip=True)
        numero = spans[1].get_text(strip=True) # contador
        tipo = spans[2].get_text(strip=True) #tipo : pasteleria, cafeteria etc
        precio = spans[3].get_text(strip=True)
        # print(localizacion, tipo, precio)

"""CREACION DE UN DATAFRAME PARA GUARDAR LOS DATOS"""
DataFrame = pd.DataFrame(columns=["Nombre", "Localizacion", "Numero", "Tipo", "Precio"]) #creacion del dataframe,  columnas

for resto, ficha in zip(resto_nombre, resto_ficha): #zip recorre dos listas a la vez, en este caso resto_nombre y resto_ficha
    nombre = resto.find("span").get_text(strip=True) #.find busca la primera etiqueta span dentro de resto y .get_text(strip=True) saca el texto sin espacios al principio y al final
    spans = ficha.find_all("span")

    if len(spans) >= 4:

        localizacion = spans[0].get_text(strip=True)
        numero = spans[1].get_text(strip=True)
        tipo = spans[2].get_text(strip=True)
        precio = spans[3].get_text(strip=True)

        DataFrame = pd.concat([DataFrame, pd.DataFrame([[nombre, localizacion, numero, tipo, precio]], columns=DataFrame.columns)], ignore_index=True)      

print(DataFrame)
```
